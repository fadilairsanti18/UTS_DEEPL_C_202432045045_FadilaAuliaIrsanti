Rating 1–2 dikategorikan sebagai sentimen negatif, sedangkan rating 4–5 sebagai sentimen positif. Rating 3 tidak digunakan. Setelah pembersihan dan penghapusan duplikasi, diperoleh 814 data untuk pemodelan LSTM.

  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">1. IMPORT LIBRARY</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat pustaka komputasi numerik, manipulasi data, visualisasi grafik, dan kerangka kerja PyTorch.</p>
</div>

```python
#202432045_FadilaAuliaIrsanti

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>

```python
#202432045_FadilaAuliaIrsanti

# 1. Muat dataset
df = pd.read_csv('Reviews_com_indodana_app.csv')

# 2. Pilih kolom yang diperlukan dan bersihkan data
df = df[['body', 'rating']].dropna().copy()
df['body'] = df['body'].astype(str).str.strip()
df = df[df['body'].str.len() > 0]

# Bentuk label biner:
# 1 = positif (rating 4-5)
# 0 = negatif (rating 1-2)
# Rating 3 dikeluarkan
df = df[df['rating'].isin([1, 2, 4, 5])].copy()
df['label'] = (df['rating'] >= 4).astype(int)

# Menghapus duplikat teks
df = df.drop_duplicates(subset='body').reset_index(drop=True)

# Case folding dan pembersihan teks
df['clean_text'] = (
    df['body']
    .str.lower()
    .str.replace(r'\s+', ' ', regex=True)
    .str.strip()
)

# 3. Membagi data train dan test secara stratified
train_df, test_df = train_test_split(
    df,
    test_size=0.20,
    random_state=45,
    stratify=df['label']
)

# Tokenisasi sederhana berdasarkan spasi
def tokenize(text):
    return text.split()

# Tokenisasi data training
train_tokens = [tokenize(text) for text in train_df['clean_text']]

# Membuat vocabulary dari data training
word_counts = {}

for tokens in train_tokens:
    for word in tokens:
        word_counts[word] = word_counts.get(word, 0) + 1

PAD_IDX = 0
UNK_IDX = 1

vocab = {
    '<PAD>': PAD_IDX,
    '<UNK>': UNK_IDX
}

for word in word_counts:
    vocab[word] = len(vocab)

# Padding sequence
MAX_LEN = 50

def encode_and_pad(text):
    tokens = tokenize(text)
    ids = [vocab.get(token, UNK_IDX) for token in tokens]

    # Memotong sequence jika lebih dari MAX_LEN
    ids = ids[:MAX_LEN]

    # Menambahkan padding jika kurang dari MAX_LEN
    ids += [PAD_IDX] * (MAX_LEN - len(ids))

    return ids

# Konversi teks menjadi sequence angka
X_train = np.array(
    [encode_and_pad(text) for text in train_df['clean_text']],
    dtype=np.int64
)

X_test = np.array(
    [encode_and_pad(text) for text in test_df['clean_text']],
    dtype=np.int64
)

y_train = train_df['label'].to_numpy(dtype=np.int64)
y_test = test_df['label'].to_numpy(dtype=np.int64)

# 4. Konversi menjadi PyTorch Tensor
X_train = torch.tensor(X_train, dtype=torch.long)
X_test = torch.tensor(X_test, dtype=torch.long)
y_train = torch.tensor(y_train, dtype=torch.long)
y_test = torch.tensor(y_test, dtype=torch.long)

# Menampilkan hasil preprocessing
print('Jumlah data setelah preprocessing:', len(df))
print('Data train:', X_train.shape, '| Data test:', X_test.shape)
print('Ukuran vocabulary:', len(vocab))
print('Distribusi label train:', torch.bincount(y_train).tolist())
print('Distribusi label test :', torch.bincount(y_test).tolist())
print('Contoh teks:', train_df.iloc[0]['body'])
print('Contoh token/padding:', X_train[0][:20].tolist())
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432045_FadilaAuliaIrsanti

class SequenceModel(nn.Module):
    def __init__(self, vocab_size, embedding_dim=64, hidden_size=64, num_classes=2):
        super(SequenceModel, self).__init__()
        self.embedding = nn.Embedding(vocab_size, embedding_dim, padding_idx=PAD_IDX)
        self.lstm = nn.LSTM(input_size=embedding_dim, hidden_size=hidden_size, batch_first=True)
        self.dropout = nn.Dropout(0.3)
        self.fc = nn.Linear(hidden_size, num_classes)

    def forward(self, x):
        embedded = self.embedding(x)
        output, (hidden, cell) = self.lstm(embedded)
        last_hidden = hidden[-1]
        out = self.dropout(last_hidden)
        return self.fc(out)

model = SequenceModel(vocab_size=len(vocab))
print(model)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432045_FadilaAuliaIrsanti

# 1. Hyperparameter
epochs = 50
learning_rate = 0.001
batch_size = 32
hidden_size = 64

# 2. DataLoader
train_dataset = TensorDataset(X_train, y_train)
test_dataset = TensorDataset(X_test, y_test)

train_loader = DataLoader(
    train_dataset,
    batch_size=batch_size,
    shuffle=True)

test_loader = DataLoader(
    test_dataset,
    batch_size=batch_size,
    shuffle=False)

# Menghitung class weight
class_counts = torch.bincount(
    y_train,
    minlength=2
).float()

class_weights = class_counts.sum() / (
    2 * class_counts)

# 3. Inisiasi model
model = SequenceModel(
    vocab_size=len(vocab),
    embedding_dim=64,
    hidden_size=hidden_size,
    num_classes=2)

# Fungsi loss
criterion = nn.CrossEntropyLoss(
    weight=class_weights)

# Optimizer
optimizer = optim.Adam(
    model.parameters(),
    lr=learning_rate)

# 4. Menyimpan loss
train_losses = []

# 5. Training Loop
for epoch in range(epochs):

    model.train()
    running_loss = 0.0

    for X_batch, y_batch in train_loader:
        optimizer.zero_grad()

        # Forward pass
        outputs = model(X_batch)

        # Hitung loss
        loss = criterion(outputs, y_batch)

        # Backward pass
        loss.backward()

        # Gradient clipping untuk stabilitas training LSTM
        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=1.0)

        # Update parameter
        optimizer.step()
        running_loss += (
            loss.item() * X_batch.size(0))

    # Rata-rata loss
    epoch_loss = (
        running_loss /
        len(train_loader.dataset))

    train_losses.append(epoch_loss)

    # Tampilkan loss setiap 5 epoch
    if (epoch + 1) % 5 == 0 or epoch == 0:

        print(
            f"Epoch [{epoch + 1:02d}/{epochs}] "
            f"- Loss: {epoch_loss:.4f}")

# 6. Evaluasi data testing
model.eval()

all_preds = []
all_probs = []
all_true = []

with torch.no_grad():

    for X_batch, y_batch in test_loader:

        logits = model(X_batch)

        probs = torch.softmax(
            logits,
            dim=1
        )[:, 1]

        preds = torch.argmax(
            logits,
            dim=1)

        all_preds.extend(
            preds.numpy())

        all_probs.extend(
            probs.numpy())

        all_true.extend(
            y_batch.numpy())

# 7. Menghitung akurasi
accuracy = accuracy_score(
    all_true,
    all_preds)

print(f"\nTest Accuracy: {accuracy:.4f}")
print("\nClassification Report:")
print(classification_report(
        all_true,
        all_preds,
        target_names=[
            "Negative",
            "Positive" ],
        digits=4
    )
)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">5. VISUALISASI LOSS TRAINING & PERBANDINGAN PREDIKSI VS AKTUAL</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menampilkan grafik konvergensi nilai Loss terhadap Epoch dan grafik evaluasi perbandingan hasil prediksi model terhadap data aktual.</p>
</div>

```python
#202432045_FadilaAuliaIrsanti

# 1. Plot Grafik Penurunan Loss Training terhadap Epoch
plt.figure(figsize=(10, 4))

plt.plot(
    range(1, len(train_losses) + 1),
    train_losses,
    linewidth=2,
    label='Training Loss')

plt.title('Grafik Penurunan Loss Selama Pelatihan LSTM')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()

# 2. Plot Perbandingan Data Aktual vs Hasil Prediksi Model
plt.figure(figsize=(12, 5))
plot_n = min(100, len(all_true))

plt.plot(
    range(plot_n),
    np.array(all_true[:plot_n]),
    marker='o',
    linewidth=1,
    label='Aktual')

plt.plot(
    range(plot_n),
    np.array(all_preds[:plot_n]),
    marker='x',
    linewidth=1,
    label='Prediksi')

plt.yticks(
    [0, 1],
    ['Negative', 'Positive'])

plt.title(
    'Perbandingan Sentimen Aktual vs Prediksi '
    '(100 Data Uji Pertama)')

plt.xlabel('Indeks Data Uji')
plt.ylabel('Kelas Sentimen')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()


# 3. Confusion Matrix
cm = confusion_matrix(all_true, all_preds)

plt.figure(figsize=(5, 4))
plt.imshow(cm)
plt.title('Confusion Matrix')
plt.xlabel('Prediksi')
plt.ylabel('Aktual')

plt.xticks(
    [0, 1],
    ['Negative', 'Positive'])

plt.yticks(
    [0, 1],
    ['Negative', 'Positive'])

for i in range(2):
    for j in range(2):
        plt.text(
            j,
            i,
            cm[i, j],
            ha='center', va='center')

plt.colorbar()
plt.show()
```
